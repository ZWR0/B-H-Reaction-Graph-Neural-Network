# Generated Fingerprint

```python
import pandas as pd
import numpy as np
from rdkit import Chem
import os
os.chdir('D:/HuaweiMoveData/Users/10130/Desktop/B-H Reaction Graph/Buchwald')
```

```python
from rdkit import RDLogger
from rdkit.Chem import AllChem#改
from tqdm import tqdm
# 禁用 RDKit 的日志输出
lg = RDLogger.logger()
lg.setLevel(RDLogger.CRITICAL)

# ------------------ 读取数据 ------------------
df = pd.read_csv('Buchwald.csv')

# ------------------ 参数设置 ------------------
output_dir = r'D:/HuaweiMoveData/Users/10130/Desktop/B-H Reaction Graph/修改/RDkitFP'#改
n_bits = 100

# ------------------ 处理每一列 ------------------
for col in df.columns[:5]:
    print(f"Processing column: {col}")
    fingerprints = []  # 每列单独保存

    # 获取该列所有 SMILES
    smiles_list = df[col].tolist()

    # 使用 tqdm 显示进度
    for smi in tqdm(smiles_list, desc=f'Processing {col}'):
        mol = Chem.MolFromSmiles(smi)
        fp = AllChem.RDKFingerprint(mol, maxPath=2, fpSize=n_bits)#改
        fingerprints.append(np.array(fp))
    data = pd.DataFrame(fingerprints, dtype=int)
    
    # 保存到指定路径
    output_path = f"{output_dir}/{col}_RDkit.csv" #改
    data.to_csv(output_path)
```

# R1Q7

```python
# 导入必要的库
from tqdm import tqdm
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import os

# ----------------------------
# 1. 加载目标变量 y
# ----------------------------
buchwald_path = r'D:\HuaweiMoveData\Users\10130\Desktop\B-H Reaction Graph\Buchwald\Buchwald.csv'
df_buchwald = pd.read_csv(buchwald_path)
y = df_buchwald['Output'].values / 100  # 归一化输出

# ----------------------------
# 2. 定义超参数
# ----------------------------
random_states = [1, 2, 3, 4, 42]
n_estimators_list = [100, 200, 300, 500, 1000]
max_depths = [10, 50, 100, 200, 500]

# ----------------------------
# 3. 要处理的文件列表
# ----------------------------
files = ['Additive.csv', 'Aryl_halide.csv', 'Base.csv', 'Ligand.csv', 'Product.csv']
base_path = r'D:\HuaweiMoveData\Users\10130\Desktop\B-H Reaction Graph\Buchwald'

# 存储所有结果的字典
results_dict = {}

# ----------------------------
# 4. 遍历每个文件
# ----------------------------
for file in files:
    file_path = os.path.join(base_path, file)
    print(f"正在处理文件: {file}")
    
    # 读取特征数据
    df = pd.read_csv(file_path)
    X = df.values
    
    # 创建结果存储结构：每个 (n_estimators, max_depth) 组合保存多个 random_state 的结果
    summary_data = []

    # 遍历每一对 (n_estimators, max_depth)
    for n_est in tqdm(n_estimators_list):
        for max_d in max_depths:
            r2_scores = []
            rmse_scores = []
            mae_scores = []

            # 遍历每个 random_state，训练并评估
            for rs in random_states:
                rf = RandomForestRegressor(n_estimators=n_est, max_depth=max_d, random_state=rs, n_jobs=-1)
                rf.fit(X, y)
                y_pred = rf.predict(X)

                # 计算评估指标
                r2 = r2_score(y, y_pred)
                mse = mean_squared_error(y, y_pred)
                rmse = np.sqrt(mse)
                mae = mean_absolute_error(y, y_pred)

                r2_scores.append(r2)
                rmse_scores.append(rmse)
                mae_scores.append(mae)

            # 计算均值和标准差
            r2_mean, r2_std = np.mean(r2_scores), np.std(r2_scores)
            rmse_mean, rmse_std = np.mean(rmse_scores), np.std(rmse_scores)
            mae_mean, mae_std = np.mean(mae_scores), np.std(mae_scores)

            # 格式化为 "mean ± std" 字符串，保留4位小数
            r2_str = f"{r2_mean:.4f} ± {r2_std:.4f}"
            rmse_str = f"{rmse_mean:.4f} ± {rmse_std:.4f}"
            mae_str = f"{mae_mean:.4f} ± {mae_std:.4f}"

            # 保存这一组超参数的结果
            summary_data.append({
                'n_estimators': n_est,
                'max_depth': max_d,
                'R2': r2_str,
                'RMSE': rmse_str,
                'MAE': mae_str
            })

    # 转为 DataFrame
    results_df = pd.DataFrame(summary_data)

    # 保存
    results_df.to_csv(f'D:/HuaweiMoveData/Users/10130/Desktop/B-H Reaction Graph/修改/RF_{file}', index=False)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
from scipy.signal import savgol_filter
os.chdir(r'D:\HuaweiMoveData\Users\10130\Desktop\B-H Reaction Graph')

# 读取数据
Chain = pd.read_csv(r'other_connection_performance\metrics_chain_4 (1).csv')
Cycle = pd.read_csv(r'other_connection_performance\metrics_cy_4 (1).csv')
Tree = pd.read_csv(r'other_connection_performance\metrics_tree_4 (1).csv')
Star = pd.read_csv(r'other_connection_performance\metrics_st_4 (1).csv')
Complete = pd.read_csv(r'other_connection_performance\metrics_com_4 (1).csv')
BH = pd.read_csv(r'BH-Graph-Performance\metrics_4 (1).csv')

# 模型名称
models = ['Chain', 'Cycle', 'Tree', 'Star', 'Complete', 'BH-RGNN']

# 将所有数据存储在一个字典中
data = {
    'Chain': Chain,
    'Cycle': Cycle,
    'Tree': Tree,
    'Star': Star,
    'Complete': Complete,
    'BH-RGNN': BH
}

# 创建图表
plt.figure(figsize=(8, 6))

# 定义颜色映射
colors = plt.cm.tab10.colors  # 使用 tab10 调色板（支持最多 10 种颜色）

# 遍历每个模型并绘制其 train_R2 和 test_R2 曲线
for i, model in enumerate(models):
    df = data[model]
    # 假设 CSV 文件中有 'Epoch', 'Train_R2', 'Val_R2' 列
    epochs = df['Epoch']
    train_r2 = df['Train_R2']
    test_r2 = df['Test_R2']

    # 对数据进行平滑处理
    train_r2_smooth = savgol_filter(train_r2, window_length=51, polyorder=3)
    test_r2_smooth = savgol_filter(test_r2, window_length=51, polyorder=3)
    
    if model == 'BH-RGNN':
        # 突出显示最后一组数据，用红色
        plt.plot(epochs, train_r2_smooth, label=f'{model} Train R²', alpha=0.8, color='red', linewidth=1.5, linestyle='-')
        plt.plot(epochs, test_r2_smooth, label=f'{model} Test R²', alpha=0.8, color='red', linewidth=1.5, linestyle='--')
    else:
        # 弱化其他模型的显示效果，降低透明度
        plt.plot(epochs, train_r2_smooth, label=f'{model} Train R²', color=colors[i], alpha=0.5, linewidth=1, linestyle='-')
        plt.plot(epochs, test_r2_smooth, label=f'{model} Test R²', color=colors[i], alpha=0.5, linewidth=1, linestyle='--')

# 设置图表样式
plt.xlabel('Epoch', fontsize=14)
plt.ylabel('RMSE Score', fontsize=14)
plt.legend(loc='upper right', fontsize=12)  # 图例放在右下角
plt.grid(True, linestyle='--', alpha=0.6)  # 添加网格线
plt.xlim(0, 2000)  # 设置横坐标范围为 0-2000
plt.ylim(0, 1.0)  # 设置纵坐标范围为 0-1（因为 R² 通常在 [0, 1] 范围内）

# 显示图表
plt.tight_layout()

# 保存
plt.savefig(r'D:\HuaweiMoveData\Users\10130\Desktop\B-H Reaction Graph\修改\R1Q14\rs=4.png', dpi=600, bbox_inches='tight', transparent=True)

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

Node = ['Aryl halide', 'Product', 'Base', 'Ligand', 'Additive']
Importance = [0.2110/15, 0.1915/15, 0.0968/3, 0.1390/4, 0.1948/22]
df = pd.DataFrame(columns = ['Node', 'Importance'])
df['Node'] = Node
df['Importance'] = Importance

# 设置绘图风格
sns.set(style="whitegrid")
plt.figure(figsize=(5, 5))

# 绘制带有立体效果的柱状图，因为没有分组，所以我们不需要 hue 参数
ax = sns.barplot(x='Node', y='Importance', hue='Node', data=df, palette='viridis', edgecolor='black', legend=False)

# 添加阴影以增加立体感
for p in ax.patches:
    left, bottom, width, height = p.get_bbox().bounds
    ax.add_patch(plt.Rectangle((left+width, bottom), 0.08, height, fill=True, facecolor='gray', alpha=0.4))
    ax.add_patch(plt.Rectangle((left, bottom+height), width, 0.001, fill=True, facecolor='gray', alpha=0.4))

# 设置标题和标签
plt.ylabel('Component average importance score', fontsize=18)
plt.xlabel('')

# 添加数值标签到每个柱子上，并调整文本格式
bars = ax.patches  # 获取所有的柱子
for bar in bars[:len(df)]:  # 遍历所有柱子
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height, f'{height:.2f}', 
            ha='center', va='bottom', fontsize=12, color='black',
            bbox=dict(facecolor='white', edgecolor='none', pad=0.5))

# 调整布局并显示图形
plt.tight_layout()
plt.savefig(r'D:\HuaweiMoveData\Users\10130\Desktop\B-H Reaction Graph\修改\figure7_3.png', dpi=600, bbox_inches='tight', transparent=True)
plt.show()
```
